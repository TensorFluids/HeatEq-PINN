```python
import torch
```

```python
x = torch.linspace(-2, 2, 11).reshape(-1,1)
```

```python
x.requires_grad_(True)
```

```python
y = x**3 + 2*x
```

```python
print ("x shape =", x.shape)
print ("y shape =", y.shape)
print ("x =", x)
print ("y =", y)
```

```python
dydx_calc = 3*x**2 + 2
dydx_calc.shape
```

```python
dydx = torch.autograd.grad(outputs = y,
                           inputs = x,
                           grad_outputs = torch.ones_like(y),
                           create_graph=True)[0]

dydx.shape

assert torch.allclose(dydx_calc, dydx), "Derivatives do not match!"
print("Derivatives Match!")
```
